# 02 — Standard RAG Experiment

End-to-end experiment with the standard RAG pipeline.

**Steps:**
1. Index all documents (chunk → embed → store)
2. Run sample queries
3. Inspect retrieval results
4. Evaluate answer quality

```python
import sys
sys.path.insert(0, '..')

from src.rag.pipeline import RAGPipeline
from src.utils.logging_setup import setup_logging
setup_logging('INFO')
```

```python
# Initialise and index
pipe = RAGPipeline()
n_chunks = pipe.index()
print(f'Indexed {n_chunks} chunks')
```

```python
# Query
question = 'What is Proximal Policy Optimization?'
result = pipe.query(question)

print(f'Question: {result.query}')
print(f'\nAnswer:\n{result.answer}')
print(f'\nRetrieved {len(result.retrieved_chunks)} chunks:')
for c in result.retrieved_chunks:
    print(f'  [{c.chunk_id}] score={c.score:.4f}  {c.text[:80]}…')
```

```python
# Evaluate (requires reference answer)
from src.evaluation.metrics import rouge_scores

reference = 'PPO is a policy gradient RL algorithm using a clipped surrogate objective.'
scores = rouge_scores(result.answer, reference)
print(f'ROUGE-1: {scores["rouge1"]:.3f}')
print(f'ROUGE-L: {scores["rougeL"]:.3f}')
```
