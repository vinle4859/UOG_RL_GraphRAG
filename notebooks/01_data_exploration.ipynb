# 01 — Data Exploration

Explore the ~1,000 ArXiv paper text files before feeding them into the RAG / GraphRAG pipelines.

**Goals:**
- Verify file counts and formats
- Check text length distribution
- Preview sample documents
- Identify data quality issues (empty files, encoding problems)
- Optionally: cluster papers by title / sub-field

```python
import sys
sys.path.insert(0, '..')  # allow importing from src/

from src.data.loader import load_documents
from src.utils.logging_setup import setup_logging
setup_logging('INFO')
```

```python
# Load all documents
docs = load_documents()
print(f'Total documents: {len(docs)}')
```

```python
# Text length distribution
import matplotlib.pyplot as plt

lengths = [len(d.text) for d in docs]
plt.figure(figsize=(10, 4))
plt.hist(lengths, bins=50, edgecolor='black')
plt.xlabel('Character count')
plt.ylabel('Number of papers')
plt.title('Paper Text Length Distribution')
plt.show()

print(f'Min: {min(lengths):,}  Max: {max(lengths):,}  Mean: {sum(lengths)/len(lengths):,.0f}')
```

```python
# Preview a sample document
sample = docs[0]
print(f'ID: {sample.doc_id}')
print(f'Length: {len(sample.text):,} chars')
print('---')
print(sample.text[:1000])
```

```python
# Check for empty / very short files
short_docs = [d for d in docs if len(d.text) < 100]
print(f'Documents < 100 chars: {len(short_docs)}')
for d in short_docs[:5]:
    print(f'  {d.doc_id}: {len(d.text)} chars')
```
